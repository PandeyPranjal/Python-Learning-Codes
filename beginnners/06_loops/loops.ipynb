```python
#19. Write a program that calculates the factorial of a number using a `while` loop.
num=int(input("Enter the number: "))
i=1
fact=1
while(i<=num):
    fact*=i
    i+=1
print(f"The factorial of {num} is :",fact)
```

```python
#24. Create a `while` loop that counts the number of times a user enters a number greater than 10.
count=0
while True:
    num=int(input("Enter the number: (0 for stop))"))
    if num==0:
        break
    if num>10:
        count+=1
        
print(count)
```

```python
#25. Write a program that prints all the prime numbers between 1 and 50 using a `for` loop.
for num in range(2, 51):
    is_prime=True
    for i in range(2, num):
        if num%i==0:
            is_prime=False
            break
    if is_prime:
        print(num)
        
```
