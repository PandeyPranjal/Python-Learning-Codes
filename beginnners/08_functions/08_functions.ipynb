```python
#21. How do you handle exceptions in a function?
def division(a, b):
    try:
        return a/b
    except ZeroDivisionError:
        return "can't divided by zero"
    
# division(10, 5)
division(10,0)
```

```python
#22. What is a recursive function?
#Ans. A function, which calls himself to solve a problem.
def factorial(n):
    if n==1:
        return 1
    return n* factorial(n-1)
print(factorial(5))
```

```python
#23. Write a recursive function to calculate the factorial of a number.
def recur(a):
    if a==1:
        return 1
    return a*recur(a-1)
print(recur(4))
```
