```python
# How can you design a base class so that all vehicles must implement start_engine()?

from abc import ABC, abstractmethod  #explaination in the 2nd cell
#Abstract Class
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass
    
#Child class
class Car(Vehicle):
    def start_engine(self):
        print("Car engine starts using key ignition")
class Bike(Vehicle):
    def start_engine(self):
        print("Bike engine starts using self button")
        
a=Car()
a.start_engine()
b=Bike()
b.start_engine()


########## Why This is Abstraction

# The parent class only says:
# start_engine() must exist
# But it doesn't define how.
# Each child decides the implementation.

# This forces a standard structure.



#NoTe:
# You cannot create object of abstract class.

```
