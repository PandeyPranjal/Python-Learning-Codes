```python
# Concept: Method Overloading (Python-style)
# Write a Python program to demonstrate method overloading by allowing a method to add one, two, or three numbers.

class Calc:
    def add(self, a=None, b=None, c=None):
        if a is not None and b is None and c is None:
            print(a)
        elif a is not None and b is not None and c is None:
            print(a+b)
        elif a is not None and b is not None and c is not None:
            print(a+b+c)
        else:
            print("Invalid Input")
            
a=Calc()
# a.add()  #It would give Invalid input
a.add(2, 3,4)  


#CHEAT SHEET
# Method Overloading (Python style)

# def method(a=None, b=None, c=None):

# Check which arguments are provided
# Execute corresponding logic

print("Now 2nd solution")

#BETTER APPROACH
class Calc:
    def add(self, *numbers):
        print(sum(numbers))
a=Calc()
a.add(3)
a.add(8,9)
a.add(2,3,2)
```

```python
# Concept: Method Overriding
# Write a Python program to demonstrate method overriding where a child class overrides a method of the parent class.
class Dad:
    def voice(self):
        print("Daddy says")
class Child(Dad):
    def voice(self):
        print("Child says")
        
# a=Dad()
# a.voice()
a=Child()
a.voice()


#VISUAL UNDERSTANDING


# Dad
#  |
#  | voice() → "Daddy says.."
#  |
# Child
#  |
#  | voice() → "Child says"
```
