```python
#2. Write a Python program to create a class Student with a constructor that accepts name and age.
# When an object is created, display the student's name and age.

class Student():
    def __init__(self, name,age):  #This is a parameterized constructor because it receives values (name, age) when the object is created.
        self.name=name
        self.age=age
        self.show_student()
    def show_student(self):
        print("Student name is "+" =>", self.name)
        print("Student age is :",self.age)
        
name=input("Enter your name: ")
age=input("Enter your age: ")
stu=Student(name,age)


#Program Flow:
# User enters name and age
#         ↓
# Object created
#         ↓
# Constructor runs
#         ↓
# Data stored in object
#         ↓
# show_student() called
#         ↓
# Details printed
```
