```python
# Write a Python program to demonstrate single inheritance where a child class inherits methods from a parent class.
class Parent:
    def det_parent(self):
        print("I am parent")
class Children(Parent):
    def det_child(self):
        print("I am Child")
        
# a=Parent()
# a.det_parent()
# a.det_child()  this will give an error because Child → can access Parent methods
                                                        # Parent → cannot access Child methods
                        
                        
a=Children()
a.det_parent()
a.det_child()                                
                                                        
# Think of it like this:

# Parent knowledge
#       ↓
# Child inherits it
# But parent doesn't inherit from child.
```

```python
#Single Inheritance
# Write a Python program to demonstrate single inheritance where a Dog class inherits a method from the Animal class.

class Animal:
    def eat(self):
        print("Animals eat")
class Dog(Animal):
    def walk(self):
        print("All the Dogs walk")
        
d=Dog()
d.walk()
d.eat()



#Visual Understanding
# Animal (Parent)
#    |
#    |--- eat()
#    |
# Dog (Child)
#    |
#    |--- walk()
```

```python
#Multiple Inheritance
#Write a Python program to demonstrate multiple inheritance where a child class inherits methods from two parent classes.
class Father:
    def dad(self):
        print("I am dad")
class Mother:
    def mom(self):
        print("I am mother")
class Children(Father, Mother):
    def child(self):
        print(" I am Children")
a=Children()
a.child()
a.dad()
a.mom()

#VISUAL UNDERSTANDING
# Papa        Mother
#    \        /
#     \      /
#      Child
```

```python
#Multilevel Inheritance
#Write a Python program to demonstrate multilevel inheritance where a class inherits from another class, which itself inherits from another class.
class Father:
    def fath(self):
        print("I am dad")
class Mother(Father):
    def mom(self):
        print("I am mother")
        
class Children(Mother):
    def child(self):
        print("I am children")
        
a=Children()
a.child()
a.mom()
a.fath()


#SYNTAX
# class A:
#     ...

# class B(A):
#     ...

# class C(B):
#     ...
```

```python
# SUPER() with Single Inheritance
# Write a Python program to demonstrate the use of super() to call a parent class method from a child class.
class Student:
    def show(self):
        print("Hii I was called using super keyword and single inheritance")
class Get(Student):
    def player(self):
        super().show()       #super() means:  go to the parent class
        print("Hii")
    
a=Get()
a.player()


#PROGRAM FLOW:

# Student (Parent)
#    |
#    |--- show()
#    |
# Get (Child)
#    |
#    |--- player()
#           |
#           |--- super().show()
```

```python
# Using super() with constructor inheritance

# Write a Python program where a child class calls the parent class constructor using super() and displays the inherited data.   


class Student:
    def __init__(self, name , age):
        self.name=name
        self.age=age
class StudentShow(Student):
    def __init__(self, name, age):
        super().__init__(name, age)   #super() → calls parent class method or constructor
        print(f"name {self.name}")
        print(f"age: {self.age}")
a=StudentShow("Pranjal", 18)



#PROGRAM FLOW:

# Object created
#       ↓
# Child constructor runs
#       ↓
# super() calls parent constructor
#       ↓
# name and age stored
#       ↓
# child prints values
```

```python
#Concept: Multilevel Inheritance + super()
# Write a Python program demonstrating multilevel inheritance where a child class inherits properties from its parent class and calls parent methods using super().

class Employee:
    def __init__(self, name, salary):
        self.name=name
        self.salary=salary
class Manager(Employee):
    def __init__(self, name, salary,dept):
        super().__init__(name, salary)
        self.dept=dept
    def show(self):
        print(f"The name {self.name}")
        print(f"The salary {self.salary}")
        print(f"The dept {self.dept}")
        
class Hr(Manager):
    def show_data(self):
        super().show()
        
a=Hr("Pranjal", 234233, "IT")
a.show_data()


#PROGRAM FLOW:
# Hr object created
#         ↓
# Employee constructor runs
#         ↓
# Manager constructor runs
#         ↓
# Hr method called
#         ↓
# Manager show() method runs
```

```python
# Concept: Hierarchical Inheritance
# Write a Python program to demonstrate hierarchical inheritance where two child classes inherit methods from the same parent class.


class Father:
    def dad(self):
        print("Wake Up!")
class elderbro(Father):
    def elder(self):
        print("I am Elder Brother")
class youngerbro(Father):
    def young(self):
        print("I am younger")
        
a=youngerbro()
a.dad()
b=elderbro()
b.elder()



#HIERARCHICAL STRUCTURE:

#         Papa
#        /    \
# BadaBhai   ChotaBhai


#HIERARCHICAL INHERITANCE SYNTAX:
# class Parent:
#     ...

# class Child1(Parent):
#     ...

# class Child2(Parent):
    # ...
```

```python
# Concept: Hybrid / Hierarchical Inheritance with super()

# Write a Python program where a Person class is inherited by both Student and Teacher classes.
# Use super() to initialize parent class attributes and display the details of both student and teacher.

class Person:
    def __init__(self, name , age):
        self.name=name
        self.age=age
    def info(self):
        print(f"The name is {self.name} and the age is {self.age}")
class Student(Person):
    def __init__(self, name, age , fee, rollNo):
        super().__init__(name, age)    # super() Used to call parent class constructor or method.
        self.fee=fee
        self.rollNo=rollNo
    def stuInfo(self):
        super().info()
        print(f"Age is {self.age} and roll No is {self.rollNo}")
        
class Teacher(Person):
    def __init__(self, name, age, dept,salary):
        super().__init__(name,age)
        self.dept=dept
        self.salary=salary
    def teacherInfo(self):
        super().info()
        print(f"The dept is {self.dept} and salary is {self.dept}")
        
a=Teacher("Pranjal", 22, "CS", 2300000)
a.teacherInfo()
b=Student("Sandhya",20, 23343, 112)
b.stuInfo()

#Cheatsheet
# Hierarchical Inheritance

# Parent
#   |
#   |---- Child1
#   |
#   |---- Child2
```
