```python
#2. Write a Python program to demonstrate instance variable and class variable.
class Student:
    school="ABC school"
    def __init__(self, name):
        self.name=name
s1=Student("Pranjal")
s2=Student("Sandhya")
print(s1.name,s1.school)
print("s2.name", s2.school)


# Note:

# Instance variable → each object has its own
# Class variable → all objects share it  


            # or  
# instance → personal data
# class → shared data
```
