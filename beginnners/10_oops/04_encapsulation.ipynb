```python
#1. Write a Python program to create a class Student where the student's name is stored as a private variable.
# Provide methods to set the name and get the name.
class Student:
    def get_name(self):  #Getter Method:This method returns the stored name.
        return self.__name
    def set_name(self, name):   #Setter Method:This method stores the name inside the object.
        self.__name=name
s=Student()
s.set_name("Pranjal")
print(s.get_name())
    
    
    
    #Note:
    
# Student Object
# __name  → hidden variable



# Concept: Encapsulation

# Private variable:
# self.__name

# Setter:
# set_name(value)

# Getter:
# get_name()
```

```python
#2. Write a Python program to create a class Employee where the salary is stored as a private variable.
# Provide methods to set the salary, get the salary, and display employee details.

class Employee:
    def __init__(self,name):
        self.name=name
        self.__salary=0
    def get_salary(self):
        return self.__salary
    def set_salary(self,amount):
        self.__salary=amount
        
    def info(self):
        print(f"The name is {self.name} and salary is {self.get_salary()}")
a=Employee("Pranjal")
a.set_salary(56000)
a.info()

#         #Program Flow
# Object created
#       ↓
# salary initialized to 0
#       ↓
# set_salary(56000)
#       ↓
# salary updated
#       ↓
# info() prints employee details
```

```python
#3. Write a Python program to create a class Student that takes a student's name and marks.
# Marks must be between 0 and 100.
# Follow encapsulation rules by keeping marks private and using getter and setter methods.


class Student:
    def __init__(self,name):
        self.name=name
        self.__marks=0  #Private varibale
    def set_marks(self, marks):
        if 0<=marks<=100:
            self.__marks=marks
        else:
            print("Marks must be between 0 and 100")
            
    def get_marks(self):
        return self.__marks
    def info(self):
        print(f"Name: {self.name}")
        print(f"Marks: {self.__marks}")
        
name=input("Enter Name: ")
marks=int(input("Enter marks (0-100): "))
s=Student(name)
s.set_marks(marks)
s.info()



# Encapsulation Example

# private variable:
# self.__marks

# setter:
# set_marks()

# getter:
# get_marks()

# validation:
# 0 <= marks <= 100
```
