```python
#4. "Write a Python program (WAP) to calculate the gross salary of an employee based on the following salary structure

# Basic Salary:
#   - ₹1 - ₹4000: HRA = 10%, DA = 50%
#   - ₹4001 - ₹8000: HRA = 20%, DA = 60%
#   - ₹8001 - ₹12000: HRA = 25%, DA = 75%
#   - More than ₹12000: HRA = 30%, DA = 80%

# The program should prompt the user to enter their basic salary and then calculate and print the gross salary"

basic_sal=float(input(" your Basic Salary: "))
if basic_sal<4000:
    hra, da=0.1, 0.5
elif basic_sal <=8000:
    hra, da=0.2, 0.6
elif basic_sal <=12000:
    hra, da=0.25, 0.75
else:
    hra, da=0.3, 0.8
gross_sal=basic_sal*(1+hra+da)
print(f"Gross salary is: {gross_sal}")
```
